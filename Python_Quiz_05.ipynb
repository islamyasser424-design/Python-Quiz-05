Case Study 1: Weekly Coffee Shop Sales Comparison
Focus Area: Stage 1 (Basic Line Plot, Figure Size, Line Styles, and Legends)

Business Context:

A local coffee shop manager wants to compare the daily coffee sales of two different branches (Branch A and Branch B) over a 5-day work week.

Dataset:
```Python
    days = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday']
    branch_a_sales = [120, 150, 130, 170, 200]
    branch_b_sales = [90, 110, 105, 140, 180]
```
Expected Output: A clean line chart showing two distinct lines mapping out the sales trend for both branches across the 5 days, complete with customized markers, grid lines, and a legend.

````python
import matplotlib.pyplot as plt

days = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday']
branch_a_sales = [120, 150, 130, 170, 200]
branch_b_sales = [90, 110, 105, 140, 180]
plt.figure(figsize=(9, 5))
plt.plot(days, branch_a_sales, color='blue', marker='o', linestyle='-', label='Branch A')
plt.plot(days, branch_b_sales, color='green', marker='s', linestyle='--', label='Branch B')
plt.title('Weekly Coffee Shop Sales')
plt.xlabel('Day')
plt.ylabel('Sales')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

Case Study 2: Daily Smartphone Battery Percentage Tracking
Focus Area: Stage 1 & 2 (Axis Limits, Custom Ticks, and Grid Customization)

Business Context:

A tech reviewer wants to track and plot a smartphone's battery drainage over specific hours of the day to evaluate power efficiency.

Dataset:
```Python
hours = [9, 11, 13, 15, 17, 19, 21]
battery_level = [100, 85, 65, 45, 30, 15, 2]

```

Expected Output: A descending line chart illustrating battery depletion from 100% down to 2%, with clear time labels on the X-axis (9 AM to 9 PM) and fixed axis boundaries.

````python
import matplotlib.pyplot as plt

hours = [9, 11, 13, 15, 17, 19, 21]
battery_level = [100, 85, 65, 45, 30, 15, 2]
plt.figure(figsize=(9, 5))
plt.plot(hours, battery_level, marker='o', color='orange')
plt.title('Smartphone Battery Level')
plt.xlabel('Time')
plt.ylabel('Battery (%)')
plt.xlim(9, 21)
plt.ylim(0, 100)
plt.xticks(hours, ['9 AM', '11 AM', '1 PM', '3 PM', '5 PM', '7 PM', '9 PM'])
plt.yticks(range(0, 101, 10))
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
````

Case Study 3: Monthly Website Traffic & Target Threshold Line
Focus Area: Stage 2 (Vertical Threshold Lines and Text Annotations)

Business Context:

A website administrator tracks monthly user visits and wants to highlight a specific month (May) where a major marketing campaign was launched, using a vertical reference line and a text note.

Dataset:

```Python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
visitors = [5000, 5300, 4800, 6000, 9500, 8800]
```
Expected Output: A line chart showing traffic growth with a prominent red dashed vertical line at May, accompanied by an arrow and text annotation pointing directly to the traffic peak.

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
visitors = [5000, 5300, 4800, 6000, 9500, 8800]
plt.figure(figsize=(9, 5))
plt.plot(months, visitors, marker='o')
# May has index 4 because counting starts at zero.
plt.axvline(x=4, color='red', linestyle='--', label='Campaign launch')
plt.annotate('Peak visits', xy=(4, 9500), xytext=(2, 10500),
             arrowprops=dict(arrowstyle='->'))
plt.ylim(0, 12000)
plt.title('Monthly Website Traffic')
plt.xlabel('Month')
plt.ylabel('Visitors')
plt.legend()
plt.tight_layout()
plt.show()
````

Case Study 4: Monthly Store Expenses & Clean Border Styling
Focus Area: Stage 2 (Removing Spines / Borders and Bar Charts)

Business Context:

An accountant wants to present monthly store expenses using a modern vertical bar chart with a minimalist aesthetic achieved by removing unnecessary top and right frame borders (spines).

Dataset:
``` Python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
expenses = [3000, 4500, 3800, 5200, 4100]
```
Expected Output: A modern red bar chart displaying monthly expenses with a clean, borderless corporate look (top and right frame lines removed).

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
expenses = [3000, 4500, 3800, 5200, 4100]
plt.figure(figsize=(8, 5))
plt.bar(months, expenses, color='red')
plt.title('Monthly Store Expenses')
plt.xlabel('Month')
plt.ylabel('Expenses')
# Hide the top and right borders.
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````
